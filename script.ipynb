```python
from pyspark.sql import SparkSession

# 1. Conecta ao Spark Connect
spark = SparkSession.builder \
    .remote("sc://localhost:15002") \
    .getOrCreate()

# 2. Cria os dados que queremos salvar
dados_usuarios = [
    ("Alice", 25, "São Paulo"),
    ("Bob", 30, "Rio de Janeiro"),
    ("Charlie", 35, "Belo Horizonte")
]
df = spark.createDataFrame(dados_usuarios, ["nome", "idade", "cidade"])

# 3. Configurações de conexão com o MySQL
# Nota: Usamos 'mysql-db' no host porque o Spark e o MySQL estão na mesma rede do Docker
properties = {
    "user": "vick_user",
    "password": "vick_password",
    "driver": "com.mysql.cj.jdbc.Driver"
}
jdbc_url = "jdbc:mysql://mysql-db:3306/meu_data_lake"

print("Enviando dados para o MySQL dentro do Docker...")

# 4. Salva o DataFrame como uma tabela no MySQL
# .mode("overwrite") substitui a tabela se ela já existir. Use "append" para adicionar dados.
df.write \
    .jdbc(url=jdbc_url, table="usuarios", mode="overwrite", properties=properties)

print("Dados salvos com sucesso na tabela 'usuarios'!")
```
