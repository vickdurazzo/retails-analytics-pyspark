```python
from pyspark.sql import SparkSession

# 1. Conecta ao Spark Connect Server que está rodando no Docker
spark = SparkSession.builder \
    .remote("sc://localhost:15002") \
    .getOrCreate()

# 2. Cria dados de teste
data = [("Alice", 25), ("Bob", 30), ("Charlie", 35)]
df = spark.createDataFrame(data, ["Nome", "Idade"])

# 3. Executa uma ação que força o processamento no Docker e traz o resultado
df.show()

print("Parabéns! O PySpark rodou com sucesso via Spark Connect!")
```
